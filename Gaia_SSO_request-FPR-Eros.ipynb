Добываем результаты наблюдений Gaia DR3 для спутников планет и астероидов. Вот <a href='https://gea.esac.esa.int/archive/documentation/GDR3/Gaia_archive/chap_datamodel/sec_dm_solar_system_object_tables/ssec_dm_sso_observation.html' target=_blanc>здесь</a> хоть какое-то описание полей базы данных.

```python
import ssl
# Это модуль для добычи данных из каталогов миссии Gaia
from astroquery.gaia import Gaia
# Это модули для удобства работы с астрономическими координатами
from astropy import units as u
from astropy.coordinates import SkyCoord
```

```python
ssl._create_default_https_context = ssl._create_unverified_context
```

```python
tables = Gaia.load_tables(only_names=True)
```

```python
for table in tables:
    print(table.name)
```

```python
# такой запрос дает первые 10 записей из базы
# query1 = """SELECT 
# TOP 10
# source_id, denomination, ra, dec
# FROM gaiadr3.sso_observation
# """

# такой запрос дает все наблюдения объекта с обозначением 
# query1 = """SELECT 
# *FROM gaiadr3.sso_observation
# WHERE number_mp = 0
# """

query1 = """SELECT 
*FROM gaiafpr.sso_source
WHERE  source_id = -4293961195
"""

job1 = Gaia.launch_job(query1)
results1 = job1.get_results()
for row in results1:
    if not row['denomination'].__contains__('Gaia'):
        print(row['denomination'])
```

```python
query1 = """SELECT 
*FROM gaiafpr.sso_observation
WHERE denomination = 'ariel'
"""
#'Oberon (UIV)'
job1 = Gaia.launch_job(query1)
results1 = job1.get_results()
for row in results1:
    print(2455197.5+row['epoch_utc'],row['ra'],row['dec'],row['ra_error_systematic'],row['dec_error_systematic'])
```
